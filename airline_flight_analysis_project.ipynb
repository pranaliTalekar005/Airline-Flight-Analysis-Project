```python

flight_count=df['Airline'].value_counts()
plt.bar(flight_count.index,flight_count.values)
plt.xlabel('Airline')
plt.ylabel('Number of Flights')
plt.title('Number of flight for each Airline')
plt.xticks(rotation=45)
plt.show()
```

```python
#average ticket price for each airline 
avg_price=df.groupby('Airline')['Ticket_Price'].mean()

#bar chart
plt.figure(figsize=(10,5))
plt.bar(avg_price.index,avg_price.values)
plt.xlabel('Airline')
plt.ylabel('Average ticket price')
plt.title('Average Ticket price by Airline')
plt.xticks(rotation=45)
plt.show()
```

```python
passenger_by_date=df.groupby('Flight_Date')['Passengers'].sum()
plt.plot(passenger_by_date.index,passenger_by_date.values,marker='o')
plt.xlabel('Flight date')
plt.ylabel('total passenger')
plt.title('total passenger by flight date')
plt.xticks(rotation=45)
plt.show()
```
